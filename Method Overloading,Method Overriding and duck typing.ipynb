```python
# Method Overriding
class Parent:
    def __init__(self, name):
        self.name = name
    def show(self):
        return f"My name is {self.name}"
class Child(Parent):
    def __init__(self, name, job):
        super().__init__(name)#Calling parent class constructor in the child class
        self.job = job
    def show(self):
        return f"I am working as a {self.job}"
c = Child("anu","student")
print(c.show())

        
    
    
    
```

```python
# method over riding through the 'super'
#parent class method executing through the 'super'
class Parent:
    def __init__(self, name):
        self.name = name
    def show(self):
        return f"My name is {self.name}"
class Child(Parent):
    def __init__(self, name, job):
        super().__init__(name)#Calling parent class constructor in the child class
        self.job = job
    def show(self):
        print(super().show())
        return f"I am working as a {self.job}"
c = Child("anu","student")
print(c.show())

        
```

```python
#method overloading 
class Math:
  def add(self, a,b):
      return a + b
  def add(self, a,b,c):
      return a + b + c
  def add(self, a,b,c,d):
      return a + b + c + d

m = Math()
print(m.add(10,20,5,6))
print(m.add(10,20,5))
print(m.add(10,20))
```

```python
#Method overloading can achieve through the default arguments or variable length of 
#positional arguments or variable length of keyword arguments
##  Method Overloading
class Math:
  
  def add(self, a,b,c=0,d=0):
      return a + b + c + d

m = Math()
print(m.add(10,20,5,6))
print(m.add(10,20,5))
print(m.add(10,20))
```

```python
class Num:
    def __init__(self, n):
        self.n = n
    #Magic Methods
    def __add__(self, other):
        return self.n + other.n
    def __sub__(self, other):
        return self.n - other.n
    def __mul__(self, other):
        return self.n * other.n
    def __truediv__(self, other):
        return self.n / other.n
    def __gt__(self, other):
        return self.n > other.n
    
    
    

n1 = Num(10)
n2 = Num(5)
print(n1.n + n2.n)
print(n1+n2)
print(n1-n2)
print(n1*n2)
print(dir(n1))
print(n1/n2)#Type Error
print(n1>n2)
```

```python
class Num:
    def __init__(self, n):
        self.n = n
    #Magic Methods
    def __add__(self, other):
        return Num(self.n + other.n)
   
    def __repr__(self):
        return f"The num Value is {self.n}"
n1 = Num(10)
n2 = Num(5)
n3 = Num(6)
n4 = n1 + n2
n5 = n4 + n3
print(n5)



```

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Point(self.x + other.x , self.y+ other.y)
    def __str__(self):
        return f"The point value is {self.x, self.y}"
    
p1 = Point(5,4)
p2 = Point(3,2)
p3 = p1 + p2
print(p3)
```
