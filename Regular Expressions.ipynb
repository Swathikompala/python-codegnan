```python
import re
string = "The sky is blue and the ocean is blue"
pat = "The"
res = re.match(pat, string)
print(res)
print(res.group())# It returns the matched pattern 
print(res.span())# It returns the matched pattern positions
```

```python
#search(pat, string):This function check entire string and it returns the first matched pattern
import re
string = "The sky is blue and the ocean is blue"
pat = "The"
res = re.search(pat, string)
print(res)
print(res.group())# It returns the matched pattern 
print(res.span())# It returns the matched pattern positions
```

```python
#find all(pat, string):This function returns all the matched patterns in the string
import re
string = "The sky is blue and the ocean is blue"
pat = "blue"
res = re.findall(pat, string)
print(res)
```

```python
##Phone number validation
import re
number =input()
pattern = r'^[6-9][0-9]{9}'
res = re.fullmatch(pattern, number)
print(res)
if res:
    print("Valid Number")
else:
    print("Not valid number")
```

```python
##Phone number validation
#-Starts with 6-9
#-All are numbers
#-Length must be 10 
import re
number =input()
pattern = r'^[6-9][0-9]{9}'
res = re.fullmatch(pattern, number)
#print(res)
if res:
    print("Valid Number")
else:
    print("Not valid number")
```

```python
##Pan card Number Validation
#-length 10
#-allow characters and digits
#-first 5 are upper case
#-next 4 are digits
#-last one is upper case
import re
number =input()
pattern = r'^[A-Z]{5}[0-4]{4}[A-Z]$'
res = re.fullmatch(pattern, number)
#print(res)
if res:
    print("Valid pan Number")
else:
    print("Not valid pan number")
```

```python
import re
number =input()
pattern = r'^[a-zA-Z0-9_.]+\@(a-zA-Z0-9)+\.[a-z]{2,3}$'
res = re.fullmatch(pattern, number)
#print(res)
if res:
    print("Valid email")
else:
    print("Not valid email")
```

```python
import re
email =input()
pattern = r'^[a-zA-Z0-9_.]+\@(a-zA-Z0-9)+\.[a-z]{2,3}[1,2]$'
res = re.fullmatch(pattern, email)
print(res)
if res:
    print("Valid email")
else:
    print("Not valid email")
```
