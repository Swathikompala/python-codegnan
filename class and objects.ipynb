```python
class Car():
    wheels = 4 #Class Attribute
    def __init__(self, color, price):
        # attributes 
        self.color = color #Instance Attribute
        self.price = price #Instance Attribute
    def __str__(self):
        return "This is car class and car color is {self.color} and the price is {self.price}"
    #Method defination
    def car_details(self):
        return f"This is car class and car color is {self.color} and price is {self.price}\
        it have 4 wheels"

#Creating an object
honda = Car('blue',1000000)
tesla = Car('red',35000000)
#Method calling
print(honda)
print(honda.car_details())
print(tesla)

```

```python
class Car():
    wheels = 4 #Class Attribute
    def __init__(self, color, price):
        # attributes 
        self.color = color #Instance Attribute
        self.price = price #Instance Attribute
   
    #Method defination
    def car_details(self):
        return f"The car is {self.color} and it's price is {self.price}"
    #class method defination
    @classmethod
    def class_method(cls, new_wheels):
        
        #Updating the class attribute
        cls.wheels = new_wheels
        return f"{new_wheels} wheels updated for car class"
    #Static Method 
    @staticmethod
    def static_method(speed):
        return f"This car moves with the speed of {speed} kmph"
        
        

#Creating an object
honda = Car('blue',1000000)
tesla = Car('red',35000000)
#Method calling

print(honda.car_details())
print(honda.wheels)
print(honda.class_method(6))
print(honda.wheels, tesla.wheels)
print(honda.static_method(120))

```
