```python

from abc import ABC, abstractmethod
class Car(ABC):
    # abstract methods
    @abstractmethod
    def engine(self):# Method declaration without implementation 
        pass
    @abstractmethod
    def tyres(self):
        pass
    @abstractmethod
    def breaks(self):
        pass

# child class
class Honda(Car):
    def engine(self):#abstract methods implementation in child's method
        return "This is honda engine"
    def tyres(self):
        return "Honda car has 4- bit tyres"
    def breaks(self):
        return "Honda car applies breakes automatically"
    def ac(self):# child class own method
        return "Honda car has centeral Ac"
        

h = Honda()
print(h.breaks())
print(h.tyres())
print(h.ac())
```
