**MULTIPLE INHERITENCE**


##Multiple inheritence :One child class getting the properties from multiple parent classes 
##it have one child class and the multiple parent classes 


```python
# Parent defination 
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
    def show(self):
        return "This is from father class"
# Parent 2 class defination 
class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
    def show(self):
        return "This is from mother class"
# Child class implementation
class Son(Father,Mother):
    def __init__(self, name, job):
        self.name = name 
        self.job = job
    def working_status(self):
        return f"I am working as a {self.job}"
s = Son("gopi","Student")
print(s.show())
print(s.working_status())
```

```python
class Father:
    def __init__(self, father_name):
        self.father_name = father_name
    def show(self):
        return "This is from father class {self.father_name}"
# Parent 2 class defination 
class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name
    def show(self):
        return "This is from mother class {self.mother_name}"
# Child class implementation
class Son(Father,Mother):
    def __init__(self, name, job, mother_name, father_name):
        Mother.__init__(self,mother_name)
        Father.__init__(self, father_name)
        self.name = name 
        self.job = job
    def working_status(self):
        return f"I am working as a {self.job}"
    def parent_details(self):
        print(Mother(self.mother_name).show())
        print(Father(self.father_name).show())
s = Son("gopi", "Student", "amma", "Nanna")
print(s.show())
print(s.working_status())
print(s.parent_details())
print(Son.mro())
```

Hierarchical Inheritence:
Multiple child classes inherite the properties from the single parent class


```python
class Fruit:
    def __init__(self, name, price):
        self.name = name 
        self.price = price
    def fruit_details(self):
        return f"The fruit name is {self.name} and the price is {self.price}"
class Banana(Fruit):
    pass
class Apple(Fruit):
    pass
b = Banana("Banana", 60)
a = Apple("Apple",150)
print(a.fruit_details())
print(b.fruit_details())
    
```

 Hybrid Inheritance :

Hybrid inheritance mixes patterns like:

Single inheritance

Multiple inheritance

Multilevel inheritance

```python
# Base class
class A:
    def show_A(self):
        print("This is class A")

# Class B inherits from A (Single / Multilevel inheritance)
class B(A):
    def show_B(self):
        print("This is class B")

# Class C also inherits from A (Hierarchical inheritance)
class C(A):
    def show_C(self):
        print("This is class C")

# Class D inherits from both B and C (Multiple inheritance)
class D(B, C):
    def show_D(self):
        print("This is class D")


# Create object of D
obj = D()

# Calling methods
obj.show_A()
obj.show_B()
obj.show_C()
obj.show_D()
print(D.mro())

```
