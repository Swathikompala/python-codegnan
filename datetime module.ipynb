```python
import datetime
print(datetime.datetime.now())#Current time stamp
print(datetime.datetime.now().time())#Current time
print(datetime.datetime.now().date())#Current date
date = datetime.datetime.now().date()
print(date.year)#It returns the year from the date
print(date.month)#It returns the month from the date
print(date.day)#It returns the day from the date
print(date.weekday())#It returns the weekday number(If monday -0...to sunday-6)
print(date.isoweekday())#It returns the weekday number(If monday -1...to sunday-7)
print(date.replace(year = 2026, month = 1))#It replaces the value 
print(date.strftime("%d - %m - %y"))
print(date.strftime("%d - %M - %y"))#M is mintues
print(date.strftime("%d / %m / %y"))
print(date.strftime("%d / %b / %y"))#b is short cut of the month
print(date.strftime("%D / %m / %y"))
print(date.strftime("%a / %m / %y"))#a is week short cut name 
#print(dir(date))
```

```python
from IPython.display import clear_output
import time

sec_counter = min_counter = hour_counter = 0
while True:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)#it waits the execution for one second
    sec_counter += 1
    if sec_counter == 60:
        min_counter += 1
        sec_counter = 0
        if min_counter == 60:
            hour_counter += 1
            min_counter = 0
    clear_output(wait = True)
  

```
